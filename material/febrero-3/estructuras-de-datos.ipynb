## Tuplas

Asociación de elementos heterogéneos

```python
t1 = ("Alice", 33)
```

```python
t2 = ("Bob", 44)
```

```python
t1
```

Una lista de tuplas

```python
visitantes = [t1, t2]
```

```python
visitantes
```

```python
# Ordenar pour algún valor de la tupla.
sorted(visitantes, key=lambda x: x[1])
```

## Diccionarios

Asociación de elementos a través de "entradas":
+ una clave
+ asociada a un valor

```python
visits = {"Alice": 80, "Bob": 93}
```

```python
visits
```

# Iterar
Recorrer una estructura de datos para hacer algo con ellos

```python
for c in mi_lista:
    print(c)
```

```python
for visitante in visitantes:
    print(visitante)
```

```python
for clave, valor in visits.items():
    print(f"Visitante {clave}")
```

```python
for clave, valor in visits.items():
    print(f"Visitante {clave} - visitas {valor}")
```

```python
# Imprimir las ciudades que contengan la letra "y"
for ciudad in mi_lista:
    if "y" in ciudad:
        print(ciudad)
```

```python
# con una "list comprehension"
[ciudad for ciudad in mi_lista if "y" in ciudad]
```

## Crear un diccionario donde las llaves sean "Alice" y "Bob" y los valores sean las ciudades que cada uno quiere visitar.
Alice quiere visitar las ciudades que contengan la letra "t" y Bob quiere visitar las ciudades que contengan la letra "h"

```python
nombres_de_ciudades = [c[0] for c in ciudades]
print(nombres_de_ciudades)
```

```python
{
    "Alice": [c for c in nombres_de_ciudades if "t" in c.casefold()],
    "Bob": [c for c in nombres_de_ciudades if "h" in c.casefold()],
}
```
