# Solución 3

La solución consisten en 3 pasos:

1. Agregar el nuevo flujo a la base de datos "biosphere3"
2. Añadir el factor de caracterización al método 
3. Agregar un enlace o arista al proceso cf modelando la nueva emisión de N2O

## Agregar nuevo flujo

```python
n2o = bd.Database('biosphere3').new_activity(
    name="monoxido dinitrogeno", 
    code='n2o', 
    categories=('air',),
    type='emission',
)
n2o.save()
```

## Añadir el factor de caracterización

```python
ipcc = bd.Method(('IPCC',))
```

### Recuperar los factores de caracterización anteriores


```python
cfs_ipcc = ipcc.load()
```

```python
print(cfs_ipcc)
```

### Agregar a los factores de caracterización el nuevo

`cfs_ipcc` es una list ade tuplas, con los factores de caracterización.
Vamos a "añadir" al final de esta lista (con el método `append` de la lista) el nuevo factor de caracterización.

```python
cfs_ipcc.append(
    (n2o.key, {'amount': 276.9})
)
```

### Escribir de nuevo el método completo
La función `write` de los objetos de la clase `Method` "sobre escribe" los factores de caracterización, por lo que necesitamos añadir a la lista CFs.

```python
ipcc.write(cfs_ipcc)
```

```python
print(cfs_ipcc)
```

## Agregar enlace o arista al proceso

```python
cf.new_exchange(
    amount=0.23, 
    type='biosphere',
    input=n2o,
).save()
```
